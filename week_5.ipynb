```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("placement_dataset.csv")


# Select only two numerical features
X = df[['IQ', 'Aptitude_Score']].copy()

# FIX: Fill missing NaN values with the column average so the scaler doesn't output NaN
X.fillna(X.mean(), inplace=True)

# Apply Standard Scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(X_scaled, columns=['IQ', 'Aptitude_Score'])

print("Original Data (First 5 rows):")
print(df[['IQ', 'Aptitude_Score']].head())

print("\nAfter Standard Scaling (NaNs Handled):")
print(scaled_df.head())
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Load the dataset
df = pd.read_csv("placement_dataset.csv")

# Select the numerical features matching your file columns
X = df[['CGPA', 'Aptitude_Score']].copy()

# Automatically clean missing values by filling them with the column average
X.fillna(X.mean(), inplace=True)

# Apply Min-Max Scaler
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Convert to a clean DataFrame
scaled_df = pd.DataFrame(X_scaled, columns=['CGPA', 'Aptitude_Score'])

print("--- PART 2: MIN-MAX SCALING ---")
print("Original Data Columns (First 5 rows):")
print(df[['CGPA', 'Aptitude_Score']].head())

print("\nAfter Min-Max Scaling (Missing Records Cleaned & Scaled):")
print(scaled_df.head())
```

```python
import pandas as pd

# Load the file you uploaded
df = pd.read_csv("placement_dataset.csv")

print("--- PART 3: DATA IMPUTATION ---")
print("Checking for missing values before cleaning:")
print(df.isnull().sum())

# Clean missing numerical values by filling them with the column average
df.fillna(df.mean(numeric_only=True), inplace=True)

# Save the clean dataset into your Colab workspace
df.to_csv("plzcementfilled.csv", index=False)

print("\nChecking for missing values after cleaning:")
print(df.isnull().sum())
print("\nNew clean dataset saved successfully as 'plzcementfilled.csv'!")
```
